```python
for image_id in image_ids:

    metadata, image = ezomero.get_image(conn=conn, image_id=image_id, dim_order='tczyx', across_groups=False, no_pixels=True)

    if metadata.getName() in os.listdir(target_folder):
        print(f'{metadata.getName()} already exists')
        continue
    
    metadata, image = ezomero.get_image(conn=conn, image_id=image_id, dim_order='tczyx', across_groups=False, no_pixels=False)
    image = image.squeeze()

    name = metadata.getName()
    os.makedirs(os.path.join(target_folder, name), exist_ok=True)
    # Save the image slices as separate files

    for z in range(image.shape[0]):
        scale = (10, float(metadata.getPixelSizeY()), float(metadata.getPixelSizeX()))
        OmeTiffWriter.save(image[z][None, :], os.path.join(target_folder, name, f"z{z}.ome.tif"), dim_order="ZYX",
                        physical_pixel_sizes=types.PhysicalPixelSizes(*scale))
    
```
