```python
from napari_autoscape.models.leica import StageOverviewRegions, CompoundShape, Point, Vertex
from dataclasses import asdict
from dicttoxml import dicttoxml
from xml.dom.minidom import parseString
from lxml import etree
```

```python
# Example: Create a minimal structure
regions = StageOverviewRegions(
    Regions=[
        CompoundShape(
            Name="J",
            Identifier="f4355853-184c-4a34-17ad-bd0b548e968d",
            Children=[
                Point(
                    Name="P1",
                    Identifier="a329381e-9932-489e-2891-cf040a4cd154",
                    Verticies=[Vertex(X=0.03912740793, Y=0.04604246217)]
                )
            ]
        )
    ],
    DefinedRegions=[
        {"Identifier": "6913fa69-69bd-4a6a-e5a9-02c4552edce6", "HolderRow": 0, "ChamberRow": 0}
    ],
    StackList=[
        {"Identifier": "a329381e-9932-489e-2891-cf040a4cd154", "Begin": 0.0013675998, "End": 0.0013675998}
    ]
)
```

```python
# Convert dataclass to dict
data = asdict(regions)

# Use dicttoxml with custom_root="StageOverviewRegions"
xml = dicttoxml(
    data,
    custom_root="StageOverviewRegions",
    attr_type=False,
    item_func=lambda x: f"Item{data['Regions'][0]['Children'].index(x) + 1}"
        if isinstance(x, dict) and 'Verticies' in x
        else None
)

# Parse and pretty-print with lxml, omitting XML declaration
root = etree.fromstring(xml)
pretty_xml = etree.tostring(
    root,
    pretty_print=True,
    encoding="utf-8",
    xml_declaration=False  # Omit the XML declaration
).decode("utf-8")

# Save to file
with open("output.xml", "w") as f:
    f.write(pretty_xml)
```
