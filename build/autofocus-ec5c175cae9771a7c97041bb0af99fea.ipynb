```python
def get_map_annotations(conn, image_id: int):
    """
    Get the map annotations for a given image ID.
    """
    # Get the image object
    map_annotation_ids = ezomero.get_map_annotation_ids(conn, object_type='Image', object_id=image_id)
    map_annotations = ezomero.get_map_annotation(conn, map_ann_id=map_annotation_ids[0])

    return map_annotations
```

```python
dataset_id = 378
```

```python
viewer.layers.clear()

image_ids = ezomero.get_image_ids(conn, dataset=dataset_id)
for image_id in tqdm.tqdm(image_ids):
    map_annotation_ids = ezomero.get_map_annotation_ids(conn, object_type='Image', object_id=image_id)
    map_annotations = ezomero.get_map_annotation(conn, map_ann_id=map_annotation_ids[0])

    metadata, image = ezomero.get_image(conn, image_id=image_id, dim_order='tczyx')

    scale=(metadata.getPixelSizeZ(), metadata.getPixelSizeY(), metadata.getPixelSizeX())
    viewer.add_image(image, name=f'Image {image_id}', translate=(map_annotations['XYStage_Y'], map_annotations['XYStage_X']), scale=scale)
```
